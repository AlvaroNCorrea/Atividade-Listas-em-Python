```python
# Jogo FizBuzz
# Objetivo: contar de 1 a 100
# Regra 1: se o numero eh divisivel por 3, escreva Fizz
# Regra 2: se o numero eh divisivel por 5, escreva Buzz
# Regra 3: se o numero eh divisivel por 3 e 5, escreva FizzBuzz
# Regra 4: se o numero nao eh divisivel por 3 ou 5, escreva o numero

for numero in range(1,101):
  if numero % 3 == 0 and numero % 5 == 0:
    print("FizzBuzz")
  elif numero % 3 == 0:
    print("Fizz")
  elif numero % 5 == 0:
    print("Buzz")
  else:
    print(numero)
```

```python
produtos = ["calca","camisa","casaco"]
precos = [10,20,30]
print("Por gentileza cadastre um novo produto e seu preco")

novo_produto = input("informe o nome do produto: ")
novo_preco = float(input("informe o preco do produto: "))

produtos.append(novo_produto)
precos.append(novo_preco)
print("produto cadastrado com sucesso")

for i,j in zip(produtos,precos):
  print(f"o produto {i} custa R${j}")

if len(produtos)>0:
  produto_removido = produtos.pop()
  preco_removido = precos.pop()
  print(f"Produto removido: {produto_removido}")
  print(f"Preço removido: R${preco_removido:.2f}")
else:
  print("Estoque vazio")
for i,j in zip(produtos,precos):
  print(f"o produto {i} custa R${j}")
```
